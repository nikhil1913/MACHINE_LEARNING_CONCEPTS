```python
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd
import warnings 
# warnings.filterwarnings('ignore')
```

```python
# adaboost classifier 
```

```python
from sklearn.datasets import make_classification
from sklearn.model_selection import train_test_split
from sklearn.ensemble import AdaBoostClassifier
from sklearn.metrics import accuracy_score, confusion_matrix , classification_report
```

```python
x , y = make_classification(n_classes=2,n_samples=1000,n_features=20,random_state=1)
```

```python
x_train , x_test , y_train , y_test = train_test_split(x,y,test_size= 0.3)
```

```python
x_train.shape , x_test.shape
```

```python
y_test.shape , y_train.shape
```

```python
ada_clf = AdaBoostClassifier()
ada_clf
```

```python
ada_clf.fit(x_train,y_train)
```

```python
y_pred = ada_clf.predict(x_test)
```

```python
print("accuraccy is ", round(accuracy_score(y_test , y_pred)*100),"%")
```

```python
print("confusion_matriix")
print(confusion_matrix(y_test,y_pred))
print("classification_report")
print(classification_report(y_test,y_pred))
```

```python
# hyper parameter tuning 
from sklearn.model_selection import GridSearchCV
params = {'n_estimators' : [50,100,150],
          'learning_rate' : [0.5,0.1,0.15],
          'algorithm' : ['SAMME']
}
gv_adaboost = GridSearchCV(estimator=ada_clf , param_grid= params , n_jobs=-1 , verbose=2,cv = 5)
```

```python
gv_adaboost.fit(x_train,y_train)
```

```python
gv_adaboost.best_estimator_
```

```python
gv_adaboost.best_score_*100
```

```python
from sklearn.datasets import make_regression
```

```python
x , y = make_regression(n_samples=1000,n_features=3,noise=10,random_state=1)
```

```python
x_train , x_test , y_train , y_test = train_test_split(x,y,test_size= 0.3)
```

```python
from sklearn.ensemble import AdaBoostRegressor

ada_reg = AdaBoostRegressor()
ada_reg
```

```python
ada_reg.fit(x_train,y_train)
```

```python
y_pred = ada_reg.predict(x_test)
```

```python
from sklearn.metrics import r2_score , mean_squared_error
print("r2_score = ",r2_score(y_test,y_pred)*100)
print("Mean_Squared_Error = ",mean_squared_error(y_test,y_pred)*100)
```

```python
# hyper parameter tuning 
from sklearn.model_selection import GridSearchCV
params = {'n_estimators' : [50,100,150],
          'learning_rate' : [0.5,0.1,0.15],
          'loss' : ['linear','square','exponential']
}
gv_adaboost_reg = GridSearchCV(estimator=ada_reg , param_grid= params , n_jobs=-1 , verbose=2,cv = 5)
```

```python
gv_adaboost_reg.fit(x_train,y_train)
```

```python
gv_adaboost_reg.best_estimator_
```

```python
gv_adaboost_reg.best_score_*100
```
